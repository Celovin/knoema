# Hierarchical Goal Pursuit

Phase 51 demonstrates a deterministic four-step research workflow: design an experiment, collect data, analyze results, and close the report loop.

```python
from knoema import AgentContext, HierarchicalPlanner, WorldState

planner = HierarchicalPlanner(default_depth=4)
tasks = planner.decompose(
    'complete a paper experiment',
    AgentContext(
        agent_id='researcher',
        location='Lab',
        active_goals=('complete a paper experiment',),
        plan_depth=4,
    ),
)
[(task.task_id, task.description, task.parent_id) for task in tasks]
```

```python
completed = []
for tick in range(4):
    task = planner.select_next_task('researcher', WorldState(tick=tick))
    if task is not None:
        completed.append(task.description)
        planner.update_progress('researcher', task)

completed
```
